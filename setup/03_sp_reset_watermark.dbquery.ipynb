```sql
CREATE OR REPLACE PROCEDURE retail.ops.reset_watermark(
    p_schema STRING DEFAULT NULL,
    p_table  STRING DEFAULT NULL
)
LANGUAGE SQL
SQL SECURITY INVOKER
AS BEGIN
    UPDATE retail.ops.watermark
    SET last_watermark = TIMESTAMP '1900-01-01 00:00:00'
    WHERE (p_schema IS NULL OR schema_name = p_schema)
      AND (p_table  IS NULL OR table_name  = p_table);
END;

CALL retail.ops.reset_watermark();                       -- reset all
CALL retail.ops.reset_watermark('silver');               -- reset all of bronze
CALL retail.ops.reset_watermark('silver', 'customers');  -- reset one table

-- DROP TABLE IF EXISTS retail.silver.orders;
-- DROP TABLE IF EXISTS retail.silver.customers;
```
