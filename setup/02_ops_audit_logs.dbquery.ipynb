```sql
CREATE SCHEMA IF NOT EXISTS retail.ops

DROP TABLE IF EXISTS retail.ops.audit_logs;

CREATE TABLE IF NOT EXISTS retail.ops.audit_logs (
    run_id STRING,
    schema_name STRING,
    table_name STRING,
    start_time STRING,
    end_time STRING,
    duration DOUBLE,
    status STRING,
    rows_written BIGINT,
    error_message STRING
)
```
