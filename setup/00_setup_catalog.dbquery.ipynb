```sql

-- 1. Create Catalog
DROP CATALOG IF EXISTS retail CASCADE;
CREATE CATALOG IF NOT EXISTS retail;

-- 2. Generate Schemas
CREATE SCHEMA IF NOT EXISTS retail.bronze;
CREATE SCHEMA IF NOT EXISTS retail.silver;
CREATE SCHEMA IF NOT EXISTS retail.gold;
CREATE SCHEMA IF NOT EXISTS retail.ops;

-- 3. Create Volume / Staging Area
CREATE VOLUME IF NOT EXISTS retail.bronze.landing;

-- 4. Create Watermark Table
CREATE TABLE IF NOT EXISTS retail.ops.watermark (
    schema_name    STRING,
    table_name     STRING,
    last_watermark TIMESTAMP
);
INSERT INTO retail.ops.watermark VALUES
    ('bronze', 'customers', TIMESTAMP '1900-01-01 00:00:00'),
    ('bronze', 'orders',    TIMESTAMP '1900-01-01 00:00:00'),
    ('silver', 'customers', TIMESTAMP '1900-01-01 00:00:00'),
    ('silver', 'orders',    TIMESTAMP '1900-01-01 00:00:00');

SELECT * FROM retail.ops.watermark;

-- 5. Create audit table
DROP TABLE IF EXISTS retail.ops.audit_logs;

CREATE TABLE IF NOT EXISTS retail.ops.audit_logs (
    run_id STRING,
    schema_name STRING,
    table_name STRING,
    start_time STRING,
    end_time STRING,
    duration DOUBLE,
    status STRING,
    rows_written BIGINT,
    error_message STRING
)
```
